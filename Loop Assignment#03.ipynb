### Question 7: ATM Menu
#### Create a simple ATM system using a while loop.
#### The initial balance is:
#### 50000
#### Display this menu repeatedly:
#### 1. Check Balance
#### 2. Deposit
#### 3. Withdraw
#### 4. Exit
#### The program should continue until the user selects 4.
#### Implement the following:
#### • Check balance
#### • Deposit money
#### • Withdraw money
#### • Prevent withdrawal if the amount is greater than the balance
#### • Exit the program

```python
balance = 50000
print("======ATM Menu======")
print("1.Check Balance")
print("2.Deposit Money")
print("3.WithDraw Money")
print("4.Exit the program")
while True:
    choice = int(input("Enter what you see in the ATM Menu"))
    if (choice == 1):
        print(f"Balance is: Rs.{balance}")
    elif(choice == 2):
        deposit_amount = int(input("Enter deposit Amount"))
        if (deposit_amount > 0):
            balance += deposit_amount
            print("Deposit Successfull")
            print(f"Updated balance: {balance}")
        else:
            print("Invalid Amount")
    elif (choice == 3):
        withdraw_amount = int(input("Enter withdraw Amount"))
        if (withdraw_amount <=0):
            print("Invalid Amount")
        elif (withdraw_amount > balance):
            print("Insufficent Balance")
        else:
            print(f"Remaining Balance: {balance - withdraw_amount}")
    elif (choice == 4):
        print("Thankyou for using ATM")
        break
    else:
        print("Invalid Choice! Please Try Again")
```

### Question 9: Student Marks Entry System
#### A teacher wants to enter student marks continuously.
#### Use a while loop.
#### The program should keep asking for marks until the teacher enters: -1 For every student's marks:
#### • 50 or above → Pass
#### • Below 50 → Fail
#### At the end, display:
#### Total Students
#### Total Passed
#### Total Failed
#### The -1 should not be counted as a student's marks.

```python
name = []
marks = {}
number = None
Total_student = 0
pass_student = 0
fail_student = 0
while number != -1:
    student = input("Enter student name")
    number = int(input(f"Enter {student} mark: ")) 
    if (number == -1):
        break
    elif (number >= 50):
        pass_student += 1
    else:
        fail_student += 1
    marks[student] = number
    Total_student +=1
print(f"Total Student = {Total_student}")
print(f"Total Pass Student = {pass_student}")
print(f"Total Fail Student = {fail_student}")
marks
```

### Question 11: Student Performance Management System

```python
number_of_Student = int(input("Enter a number of student: "))
pass_student = 0
fail_student = 0
highest_percentage = 0
lowest_percentage = 100
class_total_marks = 0
class_total_percentage = 0
for i in range(number_of_Student):
    print(f"Enter a detail of student {i + 1}")

    name_Student = input("Enter your name:")
    marks_python = int(input("Enter python marks: "))
    marks_english = int(input("Enter english marks: "))
    marks_math = int(input("Enter math marks: "))
    total_marks = marks_python + marks_english + marks_math
    percentage = (total_marks/300)*100

    if (percentage >=50 and marks_python >=40 and marks_english >= 40 and marks_math >= 40):
        result = "Pass"
        pass_student += 1
        if (percentage >= 90):
            Grade = "A+"
        elif (percentage >=80 and percentage <= 89):
            Grade = "A"
        elif (percentage >=70 and percentage <= 79):
            Grade = "B"
        elif (percentage >=60 and percentage <= 69):
            Grade = "C"
        else:
            Grade = "D"
    else:
        result = "Fail"
        Grade = "N/A"
        fail_student += 1
    print("======Student Report======")
    print(f"Name: {name_Student}")
    print(f"Total marks: {total_marks} / 300")
    print(f"Percentage: {percentage}%")
    print(f"Result: {result}")
    print(f"Grade: {Grade}")

    class_total_marks += total_marks
    class_total_percentage += percentage

    if (i == 0):
        highest_percentage = percentage
        lowest_percentage = percentage
    else:
        if (percentage > highest_percentage):
            highest_percentage = percentage
        elif (percentage< lowest_percentage):  
            lowest_percentage = percentage
if (number_of_Student > 0):
    class_avg = class_total_percentage/number_of_Student
else: 
    class_avg = 0

print("===== CLASS REPORT =====")
print(f"Total Students: {number_of_Student}")
print(f"Passed: {pass_student}")
print(f"Failed: {fail_student}")
print(f"Highest Percentage: {highest_percentage} %")
print(f"Lowest Percentage: {lowest_percentage} %")
print(f"Class Total Marks: {class_total_marks}")
print(f"Class Average:{class_avg}%")
```
